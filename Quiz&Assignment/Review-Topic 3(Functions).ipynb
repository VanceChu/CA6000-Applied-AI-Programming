### Explain the purpose of a Lambda function in Python.

A lambda function is an anonymous function in python.
It allows us to write a small function in one line without using 'def' and name.
We usually use lambda functions for simple tasks that are run only once.

```python
mac = lambda x,y:(x*y)+y
print(mac(3,4))
```

In Python, yield is used in a generator function.
When the function reaches yield, it returns one value without ending the function and pauses its state.
Each time the genarator is iterated(e.g. in a loop), the function continues from the last yield and produces the next vaule.
This makes it highly efficient for processing big data or large sequences because it reduces memory usage by not storing all values in memory at once.

```python
import sys

# 方法 A: 普通函数 (一次性把 100万个数据存进内存)
def use_list(n):
    result = []
    for i in range(n):
        result.append(i)
    return result

# 方法 B: 生成器函数 (用 yield，来一个处理一个，不存内存)
def use_generator(n):
    for i in range(n):
        yield i  # 关键点：暂停并返回

# --- 测试对比 ---
n = 1_000_000 # 模拟大数据量

# 1. 创建对象
big_list = use_list(n)       # 内存瞬间飙升
big_gen = use_generator(n)   # 内存几乎无变化

# 2. 查看内存占用 (Bytes)
print(f"List 占用内存: {sys.getsizeof(big_list)} bytes")
print(f"Generator 占用内存: {sys.getsizeof(big_gen)} bytes")
```
