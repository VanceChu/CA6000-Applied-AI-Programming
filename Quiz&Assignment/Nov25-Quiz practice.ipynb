### Boolean Logic



1. 三大基本运算符
* **`and` (与)**: 全真才为真 (True only if *both* are True).
* **`or` (或)**: 有真即为真 (True if *at least one* is True).
* **`not` (非)**: 取反 (Inverts the value).

2. 运算优先级 (Precedence) - ⚠️ 核心考点
当一行代码中有多个逻辑运算符时，Python **不是**简单地从左到右计算，而是严格遵守优先级：

> **`not` (最高) > `and` (中) > `or` (最低)**

3. 经典考题解析
针对语句 `result = True or False and False`，计算过程如下：
1.  **第一步 (先算 `and`)**: 优先级高。计算 `False and False`，结果是 `False`。
2.  **第二步 (再算 `or`)**: 表达式变成了 `True or False`。
3.  **最终结果**: `True`。

*(口诀记忆：就像数学里先乘除后加减一样，这里是先 `and` 后 `or`)*

```python
status = True or False
print(status)
```

### Briefly explain the operation of a generator function

We use keyword yield in generator function.

When program reaches yield keyword, it will return a value without exiting the function and saving the current state.

In the next iterator, program will run continue from its last yield keyword.

```python
[x+1 for x in range(7) if x % 2 == 0]
```

### Explain the logic on the program below(要把井字棋的搞懂)

```python
import random
loc = [' ']*9

def AI_turn():
    while True:
        choice=random.randint(1,9)
        if loc[choice] == ' ':
            loc[choice] = 'o'
            break
```

```python
print(random.randint(1,9))
print(loc)
```

The function is used to make a random move for the computer.
Initially, the program uses a while loop to keep trying until a vaild move is found.
Inside the while loop, it repeatedly generates a random integer between 1 and 9 to select a spot on the board.
Then, the code checks if the selected position in the list loc is empty. And if it's empty, it updates that position with 'o' and breaks the loop to end the turn. If the position is not empty, the loop runs again.
Additionally, there is a bug in the code. random.randint(1,9) generates numbers up to 9, but the list loc's indices are 0-8. This will cause an IndexError if 9 is generated, and index 0 will never be selected.

The function AI_turn attempts to make a random move for the computer player.  
It starts an infinite loop (while True) to ensure a valid move is eventually made.  
Inside the loop, it generates a random integer between 1 and 9 representing a board position.  
It checks if that position is empty (contains a space ' ').  
If the spot is empty, it assigns the value 'o' to that position and uses break to exit the loop.  
If the spot is already occupied, the loop runs again to pick a different random number.
