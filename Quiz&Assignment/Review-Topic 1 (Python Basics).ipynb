## Casting for input()

input （） function by default treats the input it receives as string

```python
x = input("Key in first number")
y = input("Key in second number")
print(x+y)
print(type(x+y))
```

which can be changed using type casting for further manipulation

```python
x = float(input("Key in first number"))
y = float(input("Key in second number"))
print(x+y)
print(type(x+y))
```

## Conditional Execution

```python
character = input("type A or B or C")
match character:
    case "A":
        print("Hhaha")
    case "B":
        print("xixixi)")
    case _: # equivalent to “else”
        print("啥也没有")
```

```python
character = input("type A or B or C")
if character == "A":
    print("hahah")
elif character == "B":
    print("xixixi")
else:
    print("啥也没有")
```
