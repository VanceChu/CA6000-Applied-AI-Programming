# Python Data Structures Comparison

| 数据结构 (Data Structure) | 有序性 (Ordered) | 可变性 (Mutable) | 允许重复 (Duplicates) | 索引访问 (Indexed) | 语法示例 (Syntax) | 核心特征/用途 (Core Features) |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| **List (列表)** | ✅ 有序 | ✅ Mutable (可变) | ✅ 允许 | ✅ 能 (`li[0]`) | `[1, 2, "a"]` | **通用容器**：可存不同类型数据 (Heterogeneous)，最常用 |
| **Tuple (元组)** | ✅ 有序 | 🔒 **Immutable (不可变)** | ✅ 允许 | ✅ 能 (`t[0]`) | `(1, 2, 3)` | **安全只读**：数据不可篡改，可作为字典 Key |
| **Dictionary (字典)** | ✅ 有序 (Py3.7+) | ✅ Mutable (可变) | ❌ Key 不行 <br> ✅ Value 允许 | ❌ 只能通过 Key | `{'name': 'Chen', 'age': 22}` | **键值映射**：Hash Map，O(1) 极速查找 |
| **Set (集合)** | ❌ 无序 | ✅ Mutable (可变) | ❌ **不允许** (自动去重) | ❌ 不能 | `{1, 2, 3}` | **数学集合**：快速去重、交集/并集运算 |
| **NumPy Array (数组)** | ✅ 有序 | ✅ Mutable (可变) | ✅ 允许 | ✅ 能 (`arr[0,1]`) | `np.array([1, 2])` | **科学计算**：必须是**同质**数据 (Homogeneous)，支持向量化运算 (Vectorization) |

| Type        | Ordered?                          | Mutable? (可修改?)          | Duplicates allowed? | 访问方式 / 特点                                              | 字面量 / 构造例子                         |
|------------|-----------------------------------|-----------------------------|---------------------|--------------------------------------------------------------|------------------------------------------|
| `list`     | ✅ 有顺序（按插入顺序）           | ✅ 可变                     | ✅ 可以有重复元素   | 按索引访问：`lst[0]`                                        | `[1, 2, 2, "a"]`                        |
| `tuple`    | ✅ 有顺序                         | ❌ 不可变                   | ✅ 可以有重复元素   | 按索引访问：`t[0]`                                          | `(1, 2, 2, "a")`                        |
| `dict`     | ✅ 有顺序（3.7+ 保留插入顺序）    | ✅ 可变                     | ❌ key 不可重复     | key→value 映射：`d["age"]`                                  | `{"name": "Chen", "age": 22}`           |
| `set`      | ❌ 无顺序                         | ✅ 可变（元素可增删）       | ❌ 不允许重复元素   | 只存唯一元素，不支持索引：不能用 `s[0]`                    | `{1, 2, 3}`                             |
| `np.array` | ✅ 有顺序（元素有固定位置 / 维度） | ✅ 可变（通常元素同一类型） | ✅ 可以有重复元素   | 类似列表按索引访问，可多维：`arr[0]`, `arr[1, 2]`，支持向量化运算 | `np.array([1, 2, 2])` / `np.array([[1,2],[3,4]])` |

### List Comprehension


List comprehension is a piece of Python syntax that allows us to create list in a concise single line 

```python
fruits = ["apple", "banana", "cherry", "durian", "pineapple"]
new_list = []
for x in fruits:
    if "a" in x:
        new_list.append(x)
print(new_list)
```

```python
fruits = ["apple", "banana", "cherry", "durian", "pineapple"]
new_list = [x for x in fruits if "a" in x]
print(new_list)
```

```python
numbers = [1,12,13,24,35,38,47]
even_numbers = [num for num in numbers if num%2 == 0]
print(even_numbers)
```

```python
files = ["data.csv", "image.png", "report.pdf", "backup.csv", "script.py"]
new_file = [x for x in files if ".csv" in x]
print(new_file)
```

```python
fruits = ["apple", "banana", "cherry", "durian", "pineapple"]
long_fruits = [x for x in fruits if len(x) > 5]
print(long_fruits)
```

## Dictionary

Dictionary allows us to index using a key
Dictionay strores data values in key:vale pairs.

```python
# Define a dictionary containing key-value pairs
car = {
    "brand": "BYD",    # The Key is "brand", the Value is "BYD"
    "model": "SEAL",   # The Key is "model", the Value is "SEAL"
    "year": 2024       # The Key is "year", the Value is 2024
}
```

```python
print(car)

# This counts how many key-value pairs exist
print(len(car))

# Access a specific value using its Key
print(car["model"])
```

```python
x = car.keys()
print(x)

x = car.values()
print(x)

print(car.items())

car_with_duplicates = {
    "brand": "BYD",
    "model": "SEAL",
    "year": 1964,  # First assignment
    "year": 2024   # Second assignment (This overwrites the previous one!)
}
print(car)
```

```python
car = {
    "brand": "BYD",    # The Key is "brand", the Value is "BYD"
    "model": "SEAL",   # The Key is "model", the Value is "SEAL"
    "year": 2015     # The Key is "year", the Value is 2024
}

car["year"] = 2025 # change value
car["color"] = "RED"# add new key and value

print(car.keys())
print(car.values())

# car.remov("model")
car.pop("model")
print(car)

# 思考一下为什么对于dictionary没有.remove
# dictionary对key的检索是o1，但是对value的是on。类比到list，就是按index进行删除的.pop()
# 以及.pop()从本质上来说其实一样
list = ["apple","apple","banana"]
list.remove("apple")
print(list)
```

### union()

```python
# --- 2. Combining Sets: Method A - union() ---
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

# Creates a NEW set. set1 and set2 remain unchanged.
set3 = set1.union(set2)

print(f"Union Result: {set3}")
```

### update()

```python
# --- 3. Combining Sets: Method B - update() ---
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

# Modifies set1 IN-PLACE.
# (Fixed Typos: 'unpdate' -> 'update')
set1.update(set2)

print(f"Updated set1: {set1}")
```

### Union, intersection, defference

```python
# 定义两个集合
Set_1 = {1, 2, 3, 4}
Set_2 = {2, 3, 5, 6}

# --- 1. Union (并集) ---
# 逻辑: 1,2,3,4 加上 2,3,5,6 -> 去重后 -> 1,2,3,4,5,6
# 符号写法 (推荐):
print(f"Union (|): {Set_1 | Set_2}") 
# 方法写法: print(Set_1.union(Set_2))


# --- 2. Intersection (交集) ---
# 逻辑: 只有 2 和 3 是两边都有的
# 符号写法 (推荐):
print(f"Intersection (&): {Set_1 & Set_2}")
# 方法写法: print(Set_1.intersection(Set_2))


# --- 3. Difference (差集) ---
# 逻辑: Set_1 有 {1,2,3,4}，减去 Set_2 {2,3,5,6}
# 2和3被减掉了，5和6本来就不在Set_1里所以不管，只剩下 1 和 4
# 符号写法 (推荐):
print(f"Difference (-): {Set_1 - Set_2}")
# 方法写法: 
print(f"Difference (.difference): {Set_1.difference(Set_2)}")
```
