### What is the value of the status in the following statment?(Boolean Logic)

````python
status = True or False
print(status)
````



1. 三大基本运算符
* **`and` (与)**: logic sum
* **`or` (或)**: logic product
* **`not` (非)**: 一元运算符

2. 运算优先级 (Precedence)
当一行代码中有多个逻辑运算符时，Python **不是**简单地从左到右计算，而是严格遵守优先级：

> **`not` (最高) > `and` (中) > `or` (最低)**  
一元>二元，product优先于sum

3. 经典考题解析
针对语句 `result = True or False and False`，计算过程如下：
1.  **第一步 (先算 `and`)**: 优先级高。计算 `False and False`，结果是 `False`。
2.  **第二步 (再算 `or`)**: 表达式变成了 `True or False`。
3.  **最终结果**: `True`。

### Fill in the blank

The data structure ____ can be used to remove duplicated elements in a list.

#### set()的作用


The data structure set can be used to remove duplicated elements in a list.

Quiz 考点提示： 如果题目问代码输出是什么，记得 Set 去重后，顺序是不保证的（虽然在处理小数字时看起来好像排好序了，但在逻辑上它是无序的）。

````python
numbers = [1, 2, 2, 3, 3, 3, 4]

print(set(numbers))

print(list(set(numbers)))

unique_numbers = list(set(numbers))  # remove duplicates using set

print(unique_numbers)
````
